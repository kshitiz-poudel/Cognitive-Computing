## Question 1: List Operations

**1.i)** Take roll number 1024170424. Extract individual digits and multiply each by 10 to form list L (10 numbers).  
Print L.

**1.ii)** Add two numbers to L — one using append() and one using insert() at specific position.  
Print L after each and explain what changed.

**1.iii)** Remove two elements from L — one using remove() and one using pop().  
Print L after each.

**1.iv)** Sort L in ascending order using sort(), then descending using sort(reverse=True).  
Print both.

**1.v)** Use slicing to print first three and last three elements of L in single line each.

**1.vi)** Use list comprehension to create new list with elements > average of L.

````python
name = "Kshitiz Poudel"
roll_number = "1024170424"
branch = "COPC"
age = 21
city = "Kathmandu"
cgpa = 7.0
````

````python

digits = [int(d) for d in roll_number]
L = [d * 10 for d in digits]

print("1.i) List L (each digit × 10):")
print(f"Digits: {digits}")
print(f"L = {L}")
````

````python

print("\n1.ii) Adding elements:")
print(f"Original L: {L}")

L.append(99)
print(f"After L.append(99): {L}")
print("  → append() adds element to end")

L.insert(2, 88)
print(f"After L.insert(2, 88): {L}")
print("  → insert() adds at index 2, shifts rest right")
````

````python

print("\n1.iii) Removing elements:")
print(f"Current L: {L}")

L.remove(99)
print(f"After L.remove(99): {L}")
print("  → remove() deletes first occurrence of value")

popped = L.pop(2)
print(f"After L.pop(2): {L}")
print(f"  → pop(2) removes element at index 2 (value was {popped})")
````

````python

print("\n1.iv) Sorting:")
print(f"Original L: {L}")

L_asc = sorted(L)
print(f"Ascending: {L_asc}")

L_desc = sorted(L, reverse=True)
print(f"Descending: {L_desc}")
````

````python

print("\n1.v) Slicing:")
print(f"List L: {L}")
print(f"First 3 (L[:3]): {L[:3]}")
print(f"Last 3 (L[-3:]): {L[-3:]}")
````

````python

print("\n1.vi) Elements greater than average:")
avg = sum(L) / len(L)
print(f"Average of L: {avg:.2f}")

greater_than_avg = [x for x in L if x > avg]
print(f"Elements > average: {greater_than_avg}")
````

## Question 2: Tuple Operations

**2)** Create tuple named scores containing 8 marks using first 8 values from list L in Q1.

**2.i)** Find highest score and its index. Find lowest score and how many times it appears.

**2.ii)** Reverse the tuple and return as list.  
Explain in one-line comment why tuples can't be reversed in-place (immutability).

**2.iii)** Ask user to input a score. Print its first occurrence index if present, else "not present" message.

**2.iv)** Attempt to change one element directly (e.g., scores[0] = 100).  
Capture error and explain in one line why this happens and how it differs from list.

**2.v)** Unpack tuple into first, second, and remaining using * operator in single line.

````python

scores = tuple(L[:8])
print("Tuple scores from first 8 elements of L:")
print(f"scores = {scores}")
````

````python

print("\n2.i) Highest and lowest scores:")
max_score = max(scores)
min_score = min(scores)
max_index = scores.index(max_score)
min_count = scores.count(min_score)

print(f"Highest score: {max_score} at index {max_index}")
print(f"Lowest score: {min_score} appears {min_count} time(s)")
````

````python

print("\n2.ii) Reversing tuple:")
reversed_scores = list(reversed(scores))
print(f"Original: {scores}")
print(f"Reversed (as list): {reversed_scores}")
print("  → Tuples immutable, so reversed() returns iterator; can't reverse in-place")
````

````python

print("\n2.iii) Search score in tuple:")
search_score = 70
print(f"Searching for: {search_score}")

if search_score in scores:
    idx = scores.index(search_score)
    print(f"Score {search_score} found at index {idx}")
else:
    print(f"Score {search_score} is not present in tuple")
````

````python

print("\n2.iv) Attempting to modify tuple element:")
print("Trying: scores[0] = 100")
try:
    scores[0] = 100
except TypeError as e:
    print(f"Error: {e}")
    print("  → Tuples immutable; lists allow assignment with same syntax, tuples don't")
````

````python

print("\n2.v) Unpacking tuple:")
first, second, *remaining = scores
print(f"first, second, *remaining = {scores}")
print(f"first: {first}")
print(f"second: {second}")
print(f"remaining: {remaining}")
````

## Question 3: Random Numbers and Prime Counting

Set random seed to roll number (1024170424) for reproducibility.

**3.i)** Generate list of 100 random numbers between 100 and 900 (inclusive).

**3.ii)** Count and print all odd numbers in the list.

**3.iii)** Count and print all even numbers in the list.

**3.iv)** Count and print all prime numbers. Additionally build list of primes using list comprehension.

**3.v)** Print number that occurs most frequently and how many times.

````python

import random

seed_value = int(roll_number)
random.seed(seed_value)

random_list = [random.randint(100, 900) for _ in range(100)]
print("3.i) Generated 100 random numbers (seed = roll number):")
print(f"Sample (first 15): {random_list[:15]}")
print(f"Total: {len(random_list)} numbers")
````

````python

print("\n3.ii) Odd numbers:")
odd_count = sum(1 for num in random_list if num % 2 == 1)
print(f"Count of odd numbers: {odd_count}")
````

````python

print("\n3.iii) Even numbers:")
even_count = sum(1 for num in random_list if num % 2 == 0)
print(f"Count of even numbers: {even_count}")
````

````python

print("\n3.iv) Prime numbers:")

def is_prime(n):
    if n < 2:
        return False
    if n == 2:
        return True
    if n % 2 == 0:
        return False
    for i in range(3, int(n**0.5) + 1, 2):
        if n % i == 0:
            return False
    return True

primes = [num for num in random_list if is_prime(num)]
print(f"Count of prime numbers: {len(primes)}")
print(f"Unique primes found: {sorted(set(primes))}")
````

````python

print("\n3.v) Most frequent number:")
from collections import Counter

freq = Counter(random_list)
most_common = freq.most_common(1)[0]
print(f"Most frequent: {most_common[0]} (appears {most_common[1]} times)")
````

## Question 4: Set Operations

Using same 8 digits of roll number (before multiplying by 10):
- Create set A = {digit × 7 for each digit}
- Create set B = {digit × 9 for each digit}

**4.vi)** Find union of A and B — all unique values across both.

**4.vii)** Find intersection of A and B — values common to both.

**4.viii)** Find A - B and B - A separately using difference().  
Explain how difference() differs from symmetric_difference().

**4.ix)** Find symmetric difference of A and B.

**4.x)** Check if A is subset of B and if B is superset of A using issubset() and issuperset().

**4.xi)** Ask user to input value X. Remove X from set A using discard().  
Explain why discard() is safer than remove().

````python

print("Set creation from roll no. digits:")
digits_only = [int(d) for d in roll_number]
print(f"Digits: {digits_only}")

A = {d * 7 for d in digits_only}
B = {d * 9 for d in digits_only}

print(f"\nSet A (digit × 7): {sorted(A)}")
print(f"Set B (digit × 9): {sorted(B)}")
````

````python

print("\n4.vi) Union:")
union = A.union(B)
print(f" {sorted(union)}")
````

````python
# 4.vii) Intersection
print("\n4.vii) Intersection :")
intersection = A.intersection(B)
print(f" {sorted(intersection)}")
````

````python

print("\n4.viii) Differences:")
diff_A_B = A.difference(B)
diff_B_A = B.difference(A)

print(f"A - B: {sorted(diff_A_B)}")
print(f"B - A: {sorted(diff_B_A)}")
print("  → difference() gives elements in one set only; symmetric_difference() gives ALL non-common")
````

````python

print("\n4.ix) Symmetric difference:")
sym_diff = A.symmetric_difference(B)
print(f"{sorted(sym_diff)}")
````

````python

print("\n4.x) Subset and superset checkings:")
is_subset = A.issubset(B)
is_superset = B.issuperset(A)

print(f"Is A subset of B? {is_subset}")
print(f"Is B superset of A? {is_superset}")
````

````python

print("\n4.xi) Remove from set A using discard():")
test_value = 7
print(f"Set A before: {sorted(A)}")
print(f"Removing: {test_value}")

A.discard(test_value)
print(f"Set A after discard({test_value}): {sorted(A)}")
print("  → discard() doesn't error if value missing; remove() raises KeyError if value doesn't exist")
````

## Question 5: Dictionary Operations

**5)** Consider dictionary with personal data:
```python
my_dict = { "name": "<your name>", "roll_no": "<your roll number>", "branch": "<your branch>", "age": <your age>, "city": "<your home city>" }

```

**5.i)** Rename key "city" to "location" without changing value (use pop() or comprehension).

**5.ii)** Add new key "cgpa" with CGPA value (7.0).

**5.iii)** Update "age" by increasing it by 1.

**5.iv)** Delete "branch" key using two methods: pop() and del.  
Explain how they differ (what does each return?).

**5.v)** Iterate using .items() and print each key-value pair as "key → value".

**5.vi)** Before accessing "email" key (doesn't exist), check with 'in' keyword and print fallback message.

**5.vii)** Create friend_dict with same keys but friend's details. Merge using {**dict1, **dict2}.  
Explain which values "win" when keys overlap.

**5.viii)** Use dict comprehension to create new dict with only string values.

````python

my_dict = {
    "name": name,
    "roll_no": roll_number,
    "branch": branch,
    "age": age,
    "city": city
}

print("5) Personal dictionary:")
for k, v in my_dict.items():
    print(f"  '{k}': '{v}'")
````

````python

print("\n5.i) Rename 'city' to 'location':")
print(f"Before: {my_dict}")
my_dict["location"] = my_dict.pop("city")
print(f"After: {my_dict}")
print("  → pop() removes 'city' key and returns its value, assigned to new 'location' key")
````

````python

print("\n5.ii) Add 'cgpa' key:")
print(f"Before: {my_dict}")
my_dict["cgpa"] = cgpa
print(f"After: {my_dict}")
````

````python

print("\n5.iii) Update age by 1:")
print(f"Age before: {my_dict['age']}")
my_dict["age"] += 1
print(f"Age after: {my_dict['age']}")
````

````python

print("\n5.iv) Delete 'branch' using two methods:")

dict_copy_1 = my_dict.copy()
print(f"\nMethod 1 - Using pop():")
print(f"  Before: {dict_copy_1}")
result = dict_copy_1.pop("branch")
print(f"  pop('branch') returned: '{result}'")
print(f"  After: {dict_copy_1}")

dict_copy_2 = my_dict.copy()
print(f"\nMethod 2 - Using del:")
print(f"  Before: {dict_copy_2}")
del dict_copy_2["branch"]
print(f"  del returns: None (no return value)")
print(f"  After: {dict_copy_2}")
print(f"\n  → pop() returns the value; del returns nothing")
````

````python

print("\n5.v) Iterate with .items():")
for key, value in my_dict.items():
    print(f"  {key} → {value}")
````

````python
print("\n5.vi) Safe access to non-existent key 'email':")
if "email" in my_dict:
    print(f"Email: {my_dict['email']}")
else:
    print("Email key not found in dictionary. Fallback: Email not provided")
````

````python

print("\n5.vii) Merge with friend's dictionary:")
friend_dict = {
    "name": "Biraj Chauhan",
    "roll_no": "1024170325",
    "branch": "COPC",
    "age": 20,
    "cgpa": 8.5
}

print(f"My dict: {my_dict}")
print(f"Friend dict: {friend_dict}")

merged = {**my_dict, **friend_dict}
print(f"Merged: {merged}")
print(f"  → When keys overlap, right dict (friend_dict) values overwrite left dict (my_dict)")
````

````python

print("\n5.viii) Dictionary comprehension - string values only:")
print(f"Original dict: {my_dict}")

string_dict = {k: v for k, v in my_dict.items() if isinstance(v, str)}
print(f"String values only: {string_dict}")
````
