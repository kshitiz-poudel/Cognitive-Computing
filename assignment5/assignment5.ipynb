Q2. Questions on Basic NumPy Array:
a) Reverse the NumPy array: arr = np.array([1, 2, 3, 6, 4, 5])
b) Find the most frequent value and their indice(s) in the following arrays:
i. x = np.array([1,2,3,4,5,1,2,1,1,1])
ii. y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

```python
arr2 = np.array([1, 2, 3, 6, 4, 5])
rev_arr = arr2[::-1]
print(rev_arr)
```

```python
x = np.array([1,2,3,4,5,1,2,1,1,1])
vals, counts = np.unique(x, return_counts=True)
freq_val = vals[np.argmax(counts)]
indices = np.where(x == freq_val)
print(freq_val)
print(indices)
```

```python
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
vals2, counts2 = np.unique(y, return_counts=True)
freq_val2 = vals2[np.argmax(counts2)]
indices2 = np.where(y == freq_val2)
print(freq_val2)
print(indices2)
```

Q4. Write program to create an 1-D NumPy array named kshitiz_poudel with evenly spaced 25 numbers from 10 to 100 using linspace(). Print the dimensions of the array, shape, total elements, the data type of each element and total number of bytes consumed by the array. Find the transpose of this array using reshape() attribute. Can we do the same with T attribute?

```python
kshitiz_poudel = np.linspace(10, 100, 25)
print(kshitiz_poudel)
```

```python
print(kshitiz_poudel.ndim)
print(kshitiz_poudel.shape)
print(kshitiz_poudel.size)
print(kshitiz_poudel.dtype)
print(kshitiz_poudel.nbytes)
```

```python
transpose_reshape = kshitiz_poudel.reshape(25, 1)
print(transpose_reshape)
```

Since kshitiz_poudel is a 1-D array, using the T attribute does not change it, as transpose has no effect on 1-D arrays. reshape() is needed to actually turn it into a column array.

```python
transpose_T = kshitiz_poudel.T
print(transpose_T)
print(transpose_T.shape)
```

Q5. Create a 2-D Array of three rows and four columns, named ucs420_kshitiz with following values – 10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35. Compute the mean, median, max, min, unique elements. Reshape the array to four rows and three columns and name it as reshaped_ucs420_kshitiz. Resize the array to two rows and three columns and name it as resized_ucs420_kshitiz.

```python
ucs420_kshitiz = np.array([[10, 20, 30, 40], [50, 60, 70, 80], [90, 15, 20, 35]])
print(ucs420_kshitiz)
```

```python
mean_val = np.mean(ucs420_kshitiz)
median_val = np.median(ucs420_kshitiz)
max_val = np.max(ucs420_kshitiz)
min_val = np.min(ucs420_kshitiz)
unique_val = np.unique(ucs420_kshitiz)

print(mean_val)
print(median_val)
print(max_val)
print(min_val)
print(unique_val)
```

```python
reshaped_ucs420_kshitiz = ucs420_kshitiz.reshape(4, 3)
print(reshaped_ucs420_kshitiz)
```

```python
resized_ucs420_kshitiz = np.resize(ucs420_kshitiz, (2, 3))
print(resized_ucs420_kshitiz)
```
